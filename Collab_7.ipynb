```python

# Tuple Operations - All in One Program

# Create a tuple
tup = (1, 2, 3, 4, 5)

# Length
print("Length:", len(tup))

# Concatenation
tup2 = (6, 7)
print("Concatenation:", tup + tup2)

# Repetition
print("Repetition:", tup * 2)

# Membership
print("Is 3 in tuple?", 3 in tup)

# Iteration
print("Elements:")
for i in tup:
    print(i)



# Indexing, Negative Indexing and Slicing
tup = ('spam', 'Spam', 'SPAM!')

print("tup[2]:", tup[2])
print("tup[-2]:", tup[-2])
print("tup[1:]:", tup[1:])

print("Using loop:")
for i in tup:
    print(i)

# Tuple Immutability and Deletion
tup = (10, 20, 30)

print("Original tuple:", tup)

# tup[0] = 100  # Not allowed (Tuples are immutable)

del tup

print("Tuple deleted")


# Tuple Functions
tup = (5, 2, 9, 1, 7)

print("Length:", len(tup))
print("Max value:", max(tup))
print("Min value:", min(tup))
```
